# 🎯 Customer Segmentation Project
### Task 2: Segmenting customers based on behavior and demographics using Machine Learning (K-Means Clustering).

```python
import pandas as pd
import numpy as np
import plotly.express as px
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

print("✅ Core data science libraries successfully imported!")
```

### 1. Ingesting & Creating Customer Demo Data
We generate a dataset of 200 mall customers containing `Age`, `Annual Income ($k)`, and `Spending Score (1-100)`.

```python
np.random.seed(42)
n_customers = 200

customer_ids = [f"CUST-{i:03d}" for i in range(1, n_customers + 1)]
ages = np.random.randint(18, 70, n_customers)
genders = np.random.choice(['Male', 'Female'], n_customers)

# Create 3 natural business behavioral clusters
income_low = np.random.normal(30, 8, 60)
spending_high = np.random.normal(80, 8, 60)

income_mid = np.random.normal(60, 10, 80)
spending_mid = np.random.normal(50, 10, 80)

income_high = np.random.normal(100, 12, 60)
spending_low = np.random.normal(25, 8, 60)

incomes = np.concatenate([income_low, income_mid, income_high])
spending_scores = np.concatenate([spending_high, spending_mid, spending_low])

# Clean boundary limits
incomes = np.clip(incomes, 15, 140)
spending_scores = np.clip(spending_scores, 1, 100)

df_customers = pd.DataFrame({
    'CustomerID': customer_ids,
    'Gender': genders,
    'Age': ages,
    'Annual Income ($k)': incomes,
    'Spending Score (1-100)': spending_scores
})

df_customers.head(10)
```

### 2. Data Preprocessing & Feature Scaling
Machine Learning algorithms work best when features are scaled to a standard scale.

```python
features = df_customers[['Annual Income ($k)', 'Spending Score (1-100)']]
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
print("📊 Features standardized and ready for AI processing.")
```

### 3. Training the K-Means AI Model
We group our customers into **3 distinct clusters** based on their shopping fingerprints.

```python
kmeans = KMeans(n_clusters=3, init='k-means++', random_state=42, n_init=10)
df_customers['Cluster_ID'] = kmeans.fit_predict(scaled_features)

# Give clusters human-readable business titles
cluster_map = {
    0: "Frugal High-Earners (Target with Luxury Deals)",
    1: "Budget Impulse Shoppers (Target with Clearance Sales)",
    2: "Average Stable Shoppers (Target with Regular Newsletters)"
}
df_customers['Customer_Segment'] = df_customers['Cluster_ID'].map(cluster_map)
df_customers[['CustomerID', 'Annual Income ($k)', 'Spending Score (1-100)', 'Customer_Segment']].head() 
```

### 4. Visualizing Customer Segments
Let's map out the groups using an interactive multi-colored scatter plot.

```python
fig = px.scatter(
    df_customers, 
    x='Annual Income ($k)', 
    y='Spending Score (1-100)',
    color='Customer_Segment',
    hover_data=['CustomerID', 'Age', 'Gender'],
    title='🛍️ AI Customer Segmentation Analysis Matrix',
    template='plotly_dark',
    color_discrete_sequence=px.colors.qualitative.Vivid
)
fig.update_traces(marker=dict(size=12, line=dict(width=1, color='White')))
fig.show()
```
