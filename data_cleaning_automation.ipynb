# ⚙️ Automated Data Cleaning & Reporting Pipeline
### Task 4: Automating the ingestion, preprocessing, cleaning, and visual reporting of inconsistent business datasets.

```python
import pandas as pd
import numpy as np
import plotly.express as px
from datetime import datetime

print("✅ Automation engine libraries loaded successfully!")
```

### 1. Generating a Messy, Broken Dataset
To simulate a real corporate scenario, we intentionally manufacture a broken dataset containing missing values (`NaN`), duplicate rows, inconsistent capitalization, and format anomalies.

```python
np.random.seed(101)
raw_data = {
    'Transaction_ID': ['TXN-001', 'TXN-002', 'TXN-003', 'TXN-003', 'TXN-004', 'TXN-005', 'TXN-006', 'TXN-007'],
    'Customer_Region': ['north', 'SOUTH', 'north', 'north', 'EAST', np.nan, 'west', 'SOUTH'],
    'Product_Category': ['Electronics', 'clothing', 'Electronics', 'Electronics', 'Home', 'Books', np.nan, 'clothing'],
    'Revenue_Amount': [1200.50, -500.00, 450.00, 450.00, np.nan, 85.20, 310.00, 150.00],
    'Quantity_Ordered': [2, 1, 1, 1, 3, np.nan, 2, 1]
}

df_messy = pd.DataFrame(raw_data)
print("🚨 Raw Messy Data Received From Source:")
df_messy
```

### 2. The Automated Cleaning Function (The Core Pipeline)
We construct a reusable, modular Python function that runs a 4-step programmatic scrub over any incoming dataset automatically.

```python
def automated_clean_pipeline(df_input):
    # Deep copy to protect historical source logs
    df = df_input.copy()
    
    # Step A: Drop absolute duplicate entries
    df = df.drop_duplicates()
    
    # Step B: Standardize text inconsistencies (Convert all text fields to neat Title Case)
    for col in ['Customer_Region', 'Product_Category']:
        df[col] = df[col].astype(str).str.strip().str.title()
    
    # Step C: Handle Missing Values / Blank Entries
    df['Customer_Region'] = df['Customer_Region'].replace('Nan', 'Unknown')
    df['Product_Category'] = df['Product_Category'].replace('Nan', 'Miscellaneous')
    
    # Fill missing quantities with the statistical median value
    df['Quantity_Ordered'] = df['Quantity_Ordered'].fillna(df['Quantity_Ordered'].median())
    
    # Step D: Correct mathematical anomalies (E.g., Negative revenue numbers or blank costs)
    df['Revenue_Amount'] = df['Revenue_Amount'].fillna(0)
    df['Revenue_Amount'] = df['Revenue_Amount'].clip(lower=0) # Resets negative outliers to 0
    
    return df

df_clean = automated_clean_pipeline(df_messy)
print("✨ Automated Data Cleaning Script Executed Successfully!")
df_clean
```

### 3. Automated Executive Reporting Summary
Our script automatically compiles data into an interactive visual report, providing leadership with actionable data instantly.

```python
# Automated calculation blocks
total_revenue_clean = df_clean['Revenue_Amount'].sum()
total_units_clean = df_clean['Quantity_Ordered'].sum()
records_processed = len(df_messy)
records_cleaned = len(df_clean)

print(f"=========================================")
print(f"📊 AUTOMATED EXECUTIVE REPORT SUMMARY")
print(f"=========================================")
print(f"- Historical Ingestion Log: {records_processed} raw entries received.")
print(f"- Production Output Log: {records_cleaned} clean rows compiled.")
print(f"- Total Validated Revenue: ${total_revenue_clean:,.2f}")
print(f"- Total Validated Volume: {int(total_units_clean)} units shipped.")
print(f"=========================================")
```

### 4. Visual Reporting Summary Output
An automated bar visualization charting clean revenue yields across geographic corporate regions.

```python
report_df = df_clean.groupby('Customer_Region').agg({'Revenue_Amount': 'sum'}).reset_index()

fig = px.bar(
    report_df, 
    x='Customer_Region', 
    y='Revenue_Amount',
    title='📈 Automated Regional Revenue Report (Post-Clean Compilation)',
    labels={'Customer_Region': 'Geographic Region', 'Revenue_Amount': 'Validated Revenue ($)'},
    template='plotly_dark',
    color='Customer_Region',
    color_discrete_sequence=px.colors.sequential.Mint
)
fig.show()
```
